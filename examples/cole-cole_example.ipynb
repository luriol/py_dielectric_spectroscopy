This notebook calculates the voltage response of a Cole–Cole dielectric to a falling voltage step (3 V to 0 V at \(t=0\)). The calculation uses a periodic square wave with a sufficiently long period to approximate an isolated step.

```python
import numpy as np
import matplotlib.pyplot as plt

from py_dielectric_spectroscopy.transform_dielectric_data import (
    V_cole_cole_sim)
```

```python
R = 1e5              # ohm
C0 = 200e-12         # F, empty cell
tau_d = 3e-3         # s
alpha = 0.8
epsilon_s = 50.0     # static permittivity
epsilon_inf = 2.0    # infinite-frequency permittivity
Delta_epsilon = epsilon_s - epsilon_inf
rho = 1e10           # sample resistivity (ohm m)

V_low, V_high = 0.0, 3.0

T = 0.03            # Square-wave period (s)
N = 10000           # Number of time points
t = (np.arange(N) +1) * T/2/N
```

```python

 V_out = V_cole_cole_sim(t, R, C0, epsilon_inf, 
        Delta_epsilon, tau_d, alpha, rho,
        V_high=V_high, V_low=V_low)
```

```python
plt.plot(t*1000,V_out)
plt.xscale('log')
plt.xlabel('time (ms)')
plt.ylabel('Voltage')
plt.title('Cole-Cole response')
```
